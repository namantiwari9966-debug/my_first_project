```python
import pandas as pd
import numpy as np
from google.colab import files
uploaded = files.upload("credit_risk_dataset001")
df = pd.read_csv("credit_risk_dataset001/credit_risk_dataset001.csv")
print("Dataset Loaded succesfully")
print("Shape,",df.shape)
```

```python
df_clean = df.copy()
print("Copy created succefully")
```

```python
print("Number of duplicate rows",df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()
print("Duplicates removed")
print("New shape",df_clean.shape)
```

```python
missing = df_clean.isnull().sum()
print("missing")
```

```python
numeric_columns = df_clean.select_dtypes(include = np.number).columns
for columns in numeric_columns:
  df_clean[columns].fillna(df_clean[columns].median())
  print("Numerical missing values handled")
```

```python
categorical_columns = df_clean.select_dtypes(include = "object").columns
for columns in categorical_columns:
  df_clean[columns].fillna(df_clean[columns].mode()[0])
  print("Categorical missing values handled")
```

```python
print(df_clean.isnull().sum().sum())
```

```python
df_clean.dtypes
```

```python
for column in df_clean.columns:
  print("\n",column)
  print(df_clean[column].unique())
```

```python
df_clean.to_csv("credit_risk_dataset001.csv",index =  False)
print("Clean datset saved succesfully")
```
